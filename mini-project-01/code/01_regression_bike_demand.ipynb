### ②-1. 시간 순서 분할 — 일부 공통 코드 + TODO

데이터를 시간의 순서대로 분할해야 합니다.
우선은 전체 데이터의 70%를 훈련 데이터로 하고 검증 데이터는 전체 날짜의 15%를 사용하고 테스트 데이터는 나머지를 테스트 데이터로 합니다.
시간의 순서대로 하기위해서 dates로 접근했습니다. (개인적으로 자전거는 추운 온도에서 타기 싫으니 단순히 날짜순이 아니라 봄,여름,가을,겨울을 고려해서 분할 하는게 어떨까 합니다.)
훈련 데이터는 12139개 검증 데이터 2616개 테스트는 2626개를 사용했습니다 

```python
# [공통] 날짜 변환 및 시간 순서 정렬
df["dteday"] = pd.to_datetime(df["dteday"])
#df["dteday"] = pd.to_datetime(df["dteday"])는 데이터 프레임 df의 dteday 열을 datetime 형식으로 변환하는 코드입니다. pd.to_datetime() 함수는 문자열 형식의 날짜를 datetime 객체로 변환합니다. 따라서 dteday 열의 값들이 날짜 형식으로 변환되어 이후 날짜 관련 연산이나 시각화에 활용가능
df = df.sort_values(["dteday", "hr"]).reset_index(drop=True)
#df = df.sort_values(["dteday", "hr"])는 데이터 프레임 df를 dteday 열과 hr 열을 기준으로 오름차순으로 정렬하는 코드입니다. reset_index(drop=True)는 정렬 후 인덱스를 재설정하고, 기존 인덱스를 제거(drop=True)합니다. 따라서 df는 날짜와 시간 순서대로 정렬된 데이터 프레임이 됩니다.
dates = np.sort(df["dteday"].unique())
#dates = np.sort(df["dteday"].unique())는 데이터 프레임 df의 dteday 열에서 고유한 날짜 값을 추출하고, 이를 오름차순으로 정렬하는 코드입니다. np.sort() 함수는 배열을 정렬하는 함수이며, df["dteday"].unique()는 dteday 열에서 중복되지 않은 고유한 날짜 값을 반환합니다. 따라서 dates에는 중복되지 않는 날짜들이 오름차순으로 정렬됩니다.
n_dates = len(dates)

# TODO 2-1. 날짜 기준 Train 약 70%, Validation 약 15%, 나머지 Test로 분할합니다.
# 같은 날짜의 행은 반드시 같은 집합에 포함되어야 합니다.
# 힌트: int(), 날짜 배열 슬라이싱, .isin()
n_train = int(n_dates*0.7)
n_valid = int(n_dates*0.15)
n_test=n_dates-n_train-n_valid
#n_train은 전체 날짜 수의 70%를 정수로 변환하여 학습 데이터의 개수를 계산합니다. n_valid는 전체 날짜 수의 15%를 정수로 변환하여 검증 데이터의 개수를 계산합니다. n_test는 전체 날짜 수에서 학습 데이터와 검증 데이터를 제외한 나머지 데이터

train_dates = dates[:n_train]
#train_dates는 전체 날짜 배열에서 처음 n_train개의 날짜를 선택하여 학습 데이터에 사용할 날짜를 지정합니다
valid_dates = dates[n_train:n_train+n_valid]
#valid_dates는 전체 날짜 배열에서 n_train번쨰부터 n_train+n_valid번째까지의 날짜를 선택하여 검증 데이터에 사용할 날짜를 지정합니다
test_dates = dates[n_train+n_valid:]
#test_dates는 전체 날짜 배열에서 n_train+n_valid번쨰부터 끝까지의 날짜를 선택하여 테스트 데이터에 사용할 날짜를 지정합니다 


train_df = df[df["dteday"].isin(train_dates)]
valid_df = df[df["dteday"].isin(valid_dates)]
test_df = df[df["dteday"].isin(test_dates)]
#train_df는 dteday 열의 값이 train_dates 배열에 포함된 행들로 구성된 데이터 프레임입니다. valid_df는 dteday 열의 값이 valid_dates 배열에 포함된 행들로 구성된 데이터 프레임입니다. test_df는 dteday 열의 값이 test_dates 배열에 포함된 행들로 구성 된 데이터입니다
```

```python
# [공통] 분할 범위 점검
for name, part in [
    ("Train", train_df),
    ("Validation", valid_df),
    ("Test", test_df),
    
]:
    
    print(name, part.shape)
    print(part["dteday"].min(), "~", part["dteday"].max())
    #dteday 열의 최소값과 최대값을 출력하여 각 데이터셋의 날짜 범위를 확인합니다 
    
```

### ②-2. 데이터 누수 확인 — TODO

이런 현상이 발생하면 데이터는 훈련 모델에 과적합을 하기 때문에 성능이 떨어지게 됩니다.

```python
# [공통] 분석에 사용할 컬럼
corr_cols = ["temp", "atemp", "hum", "windspeed", "cnt"]
#corr_cols는 상관계수를 계산할 떄 사용할 열들의 이름을 리스트로 지정합니다. 여기서는 온도(temp), 체감온도(atemp), 습도(hum), 풍속(windspeed), 대여량(cnt) 열을 선택합니다
# TODO 2-3. Train 데이터로만 상관계수를 계산합니다.
# 힌트: train_df[corr_cols].corr()
corr = train_df[corr_cols].corr()
#corr = train_df[corr_cols].corr()는 train_df 데이터 프레임에서 corr_cols에 지정된 열들만 선택한 후, corr() 메서드를 사용하여 상관계수를 계산합니다. 따라서 corr에는 train 데이터 셋에서 선택된 열들 간의 상관계수가 저장됩니다.

display(corr.round(3))
#display(corr.round(3))는 corr 데이터 프레임의 상관계수를 소수점 3자리로 반올림하여 출력합니다. 따라서 상관계수를 보기 쉽게 확인 가능
```

## ③ 특성공학 및 데이터 전처리 — CP-03

### ③-1. 출퇴근 특성 생성 — TODO

출퇴근할떄는 다른 시간대와 특성이 완전히 다르므로 따로 취급해야 합니다.
따라서 hr,workingday를 이용해 특성을 추가합니다. 이전 CheckPoint에서 한 데이터 누수와는 다르게 있지 말아야할 데이터가 훈련되는게 아니기 때문에 악영향은 없을거라고 보입니다.

```python
# [공통] 실습에서 사용할 출퇴근 시간 가정
COMMUTE_HOURS = [7, 8, 9, 17, 18, 19]


def add_commute_feature(X):
    """기존 hr와 workingday로 is_commute_hour 특성을 추가합니다."""
    X = X.copy()

    # TODO 3-1. workingday == 1이면서 hr이 COMMUTE_HOURS에 있으면 1, 아니면 0
    # 힌트: isin(), &, astype(int)
    X["is_commute_hour"] = ((X["workingday"]==1)&(X["hr"].isin(COMMUTE_HOURS))).astype(int)
    

    return X


# [공통] 모든 집합에 같은 특성공학 함수 적용
X_train = add_commute_feature(X_train)
X_valid = add_commute_feature(X_valid)
X_test = add_commute_feature(X_test)
```

```python
# [공통] 간단 검증 — 기대 결과는 [1, 0, 0, 1]
check = pd.DataFrame({"hr": [8, 8, 14, 18], "workingday": [1, 0, 1, 1]})
print(add_commute_feature(check))
```

### ③-2. 특성 유형 구분 및 ColumnTransformer — TODO

데이터의 표준화를 하기전에 생각할떄 이진 분류인지,범주형인지,수지형인지 구분해야 합니다.
시즌,일주일,계절등등 수치는 단순히 분류일뿐 실질적인 의미가 없는 형식적이기 때문에 범주형으로 하고 기온,습도,풍속은 수치가 크게 유의미하기 때문에 수치형으로 설정하고
통근시간,일하는 날들은 맞는가,틀리는가 두개의 진리값을 가지므로 이진분류로 합니다 

```python
# TODO 3-2. 선택한 입력 특성을 세 유형으로 구분합니다.
# 범주형(One-Hot 인코딩), 수치형(StandardScaler), 이진형(그대로 사용)
# ②교시 FEATURES와 새로 만든 is_commute_hour가 모두 포함되어야 합니다.
categorical_features = ["season","mnth","weekday","weathersit","hr"]
numeric_features = ["temp","hum","windspeed"]
binary_features = ["yr","holiday","workingday","is_commute_hour"]

# [공통] 중복/누락 없는지 확인
all_features = (categorical_features + numeric_features + binary_features)
assert len(all_features) == len(set(all_features)), "특성 목록이 중복됩니다."
assert set(all_features) == set(X_train.columns), "분류되지 않은 컬럼이 있습니다."

print("범주형:", categorical_features)
print("수치형:", numeric_features)
print("이진형:", binary_features)
```

범주형은 원핫인코더로 변환하고 수치형은 StandardScaler를 사용해 (실제값-평균)/표준편차로 표준화를 하고 이진형은 passthrough를 사용해 이진형으로 분류합니다 

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer


def make_preprocessor():
    # TODO 3-3. 다음 세 변환을 ColumnTransformer로 결합합니다.
    # 범주형: OneHotEncoder(handle_unknown="ignore", sparse_output=False)
    # 수치형: StandardScaler()
    # 이진형: "passthrough"
    preprocessor = ColumnTransformer(
        transformers=[
            (
                "cat",
                OneHotEncoder(
                    handle_unknown="ignore",
                    sparse_output=False
                ),
                categorical_features
            ),

            (
                "num",
                StandardScaler(),
                numeric_features
            ),

            (
                "bin",
                "passthrough",
                binary_features
            )
        ]
    )

    return preprocessor


# [공통] 전처리 객체 생성
preprocessor = make_preprocessor()

```

### ③-3. 전처리 결과 확인 — 공통 코드

## ④ 기준 모델 및 선형회귀 — CP-04
**목표:** DummyRegressor·LinearRegression을 학습하고 MAE·RMSE·R² 공통 평가 함수를 구현합니다.

### ④-1. 모델 구성과 학습 — TODO

모델을 평가할떄 보통 평균값을 이용한 모델과 비교를 많이 하기 떄문에 DummyRegressor를 이용해서 LinearRegression이 이번 회귀에 얼마나 어울리는지 검증하도록 합니다. 평가할 지표로는 MAE 평균절대오차 직관적인 해석에 좋습니다. RMSE 평균제곱오차로 오차의 합이 같더라도 표준편차가 큰지 판별이 가능합니다.
R^2는 1- 시그마 (실제값-모델의 예측값)^2 / (실제값-평균값)^2 로 값이 1이 나오면 분자의 합이 0 즉 모델의 예측값 전체가 실제값과 일치했다는 의미이고 음수가 나오면 평균보다 값이 크게 나왔다는 의미로 맞지 않는 모델을 사용했다는 의미입니다.

```python
from sklearn.pipeline import Pipeline
from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression

# TODO 4-1. 전처리 + DummyRegressor(strategy="mean") Pipeline 구현
dummy_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", DummyRegressor(strategy="mean"))
    ]
)

# TODO 4-2. 전처리 + LinearRegression() Pipeline 구현
linear_model = Pipeline(
    steps=[("preprocessor", preprocessor),("model", LinearRegression())]
)

# [공통] 모델 학습
dummy_model.fit(X_train, y_train)
linear_model.fit(X_train, y_train)
print("학습 완료")
```

### ④-2. 평가 함수 — TODO / 결과 정리 함수 — 공통

```python
from sklearn.metrics import (
    mean_absolute_error,
    root_mean_squared_error,
    r2_score,
)


def evaluate_regression(model, X, y):
    """MAE, RMSE, R2 딕셔너리를 반환합니다."""
    # TODO 4-3. model.predict(X)로 예측하고 세 지표를 계산합니다.
    pred = model.predict(X)
    return {
        "MAE": mean_absolute_error(y,pred),
        "RMSE": root_mean_squared_error(y,pred),
        "R2": r2_score(y,pred)
    }


def summarize(name, model):
    """Train·Validation 평가를 한 행으로 정리합니다 (공통)."""
    tr = evaluate_regression(model, X_train, y_train)
    va = evaluate_regression(model, X_valid, y_valid)
    return {
        "Model": name,
        "Train MAE": tr["MAE"],
        "Train RMSE": tr["RMSE"],
        "Train R2": tr["R2"],
        "Valid MAE": va["MAE"],
        "Valid RMSE": va["RMSE"],
        "Valid R2": va["R2"],
    }
```

결과적으로 모든 지표에서 DummyRegressor 보다 좋게 나왔습니다.
다만 Train MAE가 Valid MAE와 차이가 크기 때문에 단순히 수치만 본다면 과적합을 의심하게 만듭니다. 다만 자전거는 온도의 영향도 받을 수 있다는 것을 생각한다면 날짜순으로 데이터를 훈련한 것이 문제가 될것입니다. 다만 이번의 목표는 내일의 수치를 맞히는게 문제인데 데이터는 2년치 데이터로 계절별로 나누기에는 양이 부족한 편입니다. 검증,테스트 데이터를 구성한다면 미래의 데이터를 맞히는데 미래의 데이터가 들어가는 모순이 발생합니다

```python
# [공통] 이후 교시에서 결과를 누적할 리스트
results = []
results.append(summarize("DummyRegressor", dummy_model))
results.append(summarize("LinearRegression", linear_model))

# TODO 4-4. 두 모델의 성능을 DataFrame으로 비교합니다.
# 힌트: pd.DataFrame(results).round(3)
results_df = pd.DataFrame(results).round(3)
print(results_df)
```

**[CP-04] 완료 기준:** 두 모델을 학습하고 공통 평가 함수로 Train·Validation 성능을 표로 비교합니다.

## ⑤ Decision Tree 모델 개선 — CP-05
**목표:** `max_depth`를 바꿔 학습·검증 MAE를 비교하고 그래프로 표현합니다.

### ⑤-1. 깊이별 모델 학습 — TODO

결정트리는 데이터를 나누는 조건을 학습하고 그 조건에 따라 분류 합니다
학습 훈련 데이터와 정답을 이용해 데이터를 나눌 특성과 기준값을 학습니다.
예측 학습된 조건을 따라 리프 노드까지 이동하여 클래스를 예측합니다
결정 트리는 지니 불순도라는 클래스의 혼합 정도의 전후 감소가 큰 방향으로 분할 조건을 찾습니다.
너무 많이 학습해서 과적합 하지 않도록 깊이 제한을 할 수 있습니다.
depth가 3일떄는 훈련 지표도 많이 부족하기 때문에 과소적합이 의심됩니다.
depth가 6일때는 훈련 지표가 많이 개선되었고 검증 지표는 오차가 큰 부분이 있어 보이기는 합니다.
depth가 12일떄는 훈련 지표가 많이 낮아졌고 검증지표도 개선되었지만 depth가 3->6일떄와 비교하면 6->12에서 검증 지표가 많이 개선되지 못해서 과적합이 의심되는경향이 있습니다.

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.pipeline import Pipeline
# [공통] 실험 설정 및 저장 구조
depths = [3, 6, 12]
tree_models = {}
tree_results = []

# TODO 5-1~5-3. 각 깊이별 Pipeline 생성·학습·평가 결과 수집
for d in depths:
    # TODO 5-1. 전처리 + DecisionTreeRegressor(max_depth=d, random_state=42)
    model = Pipeline(steps=[("preprocessor", preprocessor),("model", DecisionTreeRegressor(max_depth=d,random_state=42))])

    # TODO 5-2. 모델 학습
    model.fit(X_train,y_train)

    # [공통] 학습한 모델 보관
    tree_models[d] = model

    # TODO 5-3. summarize()로 성능 결과 수집
    row = summarize(F"DecisionTredd_depth_{d}", model)
    tree_results.append(row)
    results.append(row)

# [공통] 결과 정리
tree_df = pd.DataFrame(tree_results)
tree_df["max_depth"] = depths
display(tree_df.round(3))
```

전부다 이전까지의 모델보다 성능이 좋습니다 여기서는 결정트리의 개수와 깊이가 더 깊은 B가 전체적으로 오차가 적은 것을 확인가능합니다.

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline
# [공통] 비교할 두 설정
rf_configs = {
    "A": {"n_estimators": 50, "max_depth": 10},
    "B": {"n_estimators": 100, "max_depth": 15},
}
rf_models = {}

# TODO 6-1~6-2. 두 설정의 Random Forest Pipeline 학습 및 성능 수집
for key, params in rf_configs.items():
    # TODO: Pipeline 구성 (RandomForestRegressor(**params, random_state=42, n_jobs=-1))
    model = Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            ("model", RandomForestRegressor(
                **params,
                random_state=33,
                n_jobs=-1
            ))
        ]
    )

    # TODO: 모델 학습
    model.fit(X_train, y_train)

    # [공통] 학습한 모델 보관
    rf_models[key] = model

    # TODO: summarize()로 결과를 results에 추가합니다.
    row = summarize(f"RandomForest_{key}", model)

    results.append(row)

    print(f"RandomForest {key} 학습 완료")
```

### ⑥-2. 전체 성능표 비교 — TODO

```python
# [공통] 지금까지 누적한 모델 결과
results_df = pd.DataFrame(results)

# TODO 6-3. Validation MAE가 낮은 순서로 정렬하고 인덱스를 정리합니다.
results_df = pd.DataFrame(results)
results_df.sort_values(
    by="Valid MAE",
    ascending=True
).reset_index(drop=True)
# TODO 6-4. Validation MAE - Train MAE를 계산합니다.
results_df["Gap (Valid-Train MAE)"] = (
    results_df["Valid MAE"] - results_df["Train MAE"]
)

display(results_df.round(3))
```

## ⑦ 오류 분석 및 최종 모델 선택 — CP-07
**목표:** 두 후보 모델의 예측 오차를 분석하고, 근거를 기록하여 최종 모델을 결정합니다.

랜덤 포레스트가 여러가지 수치면에서 우수했습니다
그나마 랜덤포레스트가 50개의 트리를 사용하고 맥스 뎁스가 10인 A의 경우 결정트리 뎁스가 12가 그나마 훈련에서 좋지만 검증 데이터에서 차이가 커서 과적합의 경향이 보이고 LinearRegression도 A에 비해 수치 차이가 큽니다.
A도 B에 비해 여러가지 수치면에서 밀리지만 GAP이 B가 A보다 차이가 커서 과적합의 경향을 보이니 그 둘의 예측 오차를 분석하고 모델을 선택할려고 합니다. 
잔차는 실제값 마이너스 평균값입니다 잔차의 경우 A가 0에서 오차가 큽니다.
전체적으로 A의 MAE가 B의 MAE보다 큽니다.
위에서 좀 GAP이 약간 차이가 나서 과적합의 경향이 B가 약간 더 높긴 하지만 성능적으로 볼떄 결정트리 100개와 최대깊이 15인 B가 더 우수하다고 생각이 되어집니다.

### ⑦-1. 후보 모델과 잔차 — TODO

```python
# TODO 7-1. results_df를 검토하여 오류 분석할 후보 모델 2개를 선택합니다.
# tree_models, rf_models 또는 앞서 학습한 모델 객체를 활용합니다.
candidates = {
    "RandomForest_A": rf_models["A"],
    "RandomForest_B": rf_models["B"]
}

# [공통] 실제값과 시간대 DataFrame
pred_df = pd.DataFrame({
    "hr": X_valid["hr"].values,
    "actual": y_valid.values,
})

# TODO 7-2. 각 모델의 Validation 예측값과 잔차를 계산합니다.
for name, model in candidates.items():
    pred_df[f"pred_{name}"] =  model.predict(X_valid)
    pred_df[f"resid_{name}"] =  (
        pred_df["actual"] - pred_df[f"pred_{name}"]
    )

display(pred_df.head())
```

```python
# [공통] 두 후보 모델을 나란히 비교합니다.
assert len(candidates) == 2, "오류 분석 후보를 2개 선택하세요."
fig, axes = plt.subplots(1, 2, figsize=(11, 4.5))
lim = pred_df["actual"].max() * 1.05

for ax, name in zip(axes, candidates):
        ax.scatter(
        pred_df["actual"],
        pred_df[f"pred_{name}"],
        alpha=0.5,
        s=15
    )

    # 실제값 = 예측값 기준선
        ax.plot(
        [0, lim],
        [0, lim],
        color="red",
        linestyle="--",
        label="Perfect Prediction"
    )

    # [공통] 축 및 제목
        ax.set_xlabel("actual cnt")
        ax.set_ylabel("predicted cnt")
        ax.set_title(name)
        ax.set_xlim(0, lim)
        ax.set_ylim(0, lim)

plt.tight_layout()
plt.show()
```

### ⑦-3. 시간대별 MAE — TODO

```python
# [공통] 시간대별 오류 DataFrame
hourly_mae = pd.DataFrame(index=range(24))
hourly_mae.index.name = "hr"

for name in candidates:
    # TODO 7-4. 모델별 절대오차와 시간대별 평균 절대오차를 계산합니다.
    abs_err = pred_df[f"resid_{name}"].abs()
    hourly_mae[name] = (
        abs_err.groupby(pred_df["hr"]).mean()
    )


display(hourly_mae.round(1))
```

```python
# TODO 7-5. 후보 모델들의 시간대별 MAE를 막대그래프로 표현합니다.
# 힌트: plt.bar(), hourly_mae
hourly_mae.plot(
    kind="bar",
    figsize=(14, 5),
    width=0.8,
    color=["steelblue", "orange"]
)

plt.title("Hourly MAE: RandomForest_A vs RandomForest_B")
plt.xlabel("Hour")
plt.ylabel("MAE")
plt.xticks(rotation=0)
plt.legend(title="Model")
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```
